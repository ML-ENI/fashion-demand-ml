# 03 — Regression EDA with development data only

This new analysis reads train and validation rows. It never plots or summarises test outcomes. The original notebook remains unchanged.


```python
USE_COLUMNS = ["split", "target_week", "sales", "target_next_week_sales"]
from pathlib import Path
import json
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "src/data/process_data.py").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("Open this notebook from fashion-demand-ml.")
sys.path.insert(0, str(ROOT))
from src.visualization.style import apply_style, add_title, add_source, display_es
apply_style()
processed = ROOT / "data/processed"
pointer = json.loads((processed / "CURRENT.json").read_text(encoding="utf-8"))
run_dir = (processed / pointer["run_dir"]).resolve()
if not run_dir.is_relative_to(processed.resolve()) or pointer.get("preview"):
    raise ValueError("CURRENT.json must select a complete publication.")
summary = json.loads((run_dir / "summary.json").read_text(encoding="utf-8"))
dataset_path = run_dir / "article_week.csv.gz"
parts = []
for chunk in pd.read_csv(dataset_path, usecols=USE_COLUMNS,
                         parse_dates=["target_week"], chunksize=250_000):
    part = chunk.loc[chunk["split"].isin(["train", "validation"])]
    if len(part):
        parts.append(part)
data = pd.concat(parts, ignore_index=True)
assert set(data["split"]) == {"train", "validation"}
print("Development rows:", len(data))
print("Test rows loaded into the analysis:", 0)
```

```python
train = data.loc[data["split"].eq("train")].copy()
validation = data.loc[data["split"].eq("validation")].copy()
target = "target_next_week_sales"
overview = data.groupby("split")[target].agg(["size", "mean", "median", "std"])
display_es(overview.round(3))
fig, ax = plt.subplots(figsize=(10, 4.5), layout="constrained")
weekly = data.groupby(["split", "target_week"])[target].mean().reset_index()
for split, color in [("train", "#6F4EAF"), ("validation", "#E17C05")]:
    subset = weekly.loc[weekly["split"].eq(split)]
    ax.plot(subset["target_week"], subset[target], color=color, label=split)
add_title(fig, "Mean next-week sales across development weeks")
ax.set_xlabel("Target week", fontweight="bold")
ax.set_ylabel("Mean recorded units", fontweight="bold")
ax.grid(axis="y", color="#DDE3EA", linewidth=.7)
ax.spines[["top", "right"]].set_visible(False)
ax.legend()
add_source(fig, "Source: H&M development partition; test excluded")
plt.show()
```

The plots describe recorded next-week sales in development data. They are not a test result and do not establish causal effects.
